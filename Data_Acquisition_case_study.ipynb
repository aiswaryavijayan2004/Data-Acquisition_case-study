Part C

Task 6

```python

gdp.rename(columns={"Country": "country"}, inplace=True)


literacy.rename(columns={"name": "country"},inplace=True)

```

```python
print(pop.columns)
print(gdp.columns)
print(internet_users.columns)
print(literacy.columns)
```

Task 7

```python
pop["population"] = pd.to_numeric(pop["population"])

gdp["GDP"] = pd.to_numeric(gdp["GDP"])

internet_users["internet_users"] = pd.to_numeric(internet_users["internet_users"])

literacy["literacy_rate"] = pd.to_numeric(literacy["literacy_rate"])
```

```python
print(pop.dtypes)
print(gdp.dtypes)
print(internet_users.dtypes)
print(literacy.dtypes)
```

Task 8

```python


df = pop.merge(gdp, on="country")
df = df.merge(internet_users, on="country")
df = df.merge(literacy, on="country")

print(df)
```

Task 9

```python
df["Internet Penetration (%)"] = (
    df["internet_users"] / df["population"] * 100
)

df["Internet Penetration (%)"] = df[
    "Internet Penetration (%)"
].round(2)
```

```python
df["GDP Per Capita"] = (
    df["GDP"] * 1000000 / df["population"]
)

df["GDP Per Capita"] = df[
    "GDP Per Capita"
].round(2)
```

```python
print(df.head())
```

Part D

Task 10

```python
import sqlite3
```

```python
conn = sqlite3.connect("country_statistics.db")
```

```python
df.to_sql(
    "country_statistics",
    conn,
    if_exists="replace",
    index=False
)
print("Database created successfully")
```

Task 11

```python
result = pd.read_sql(
    "SELECT * FROM country_statistics LIMIT 5",
    conn
)

print(result)
```
